```python
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset('flights')
df.head()
```

```python
sns.set_theme(style="whitegrid", rc ={"figure.figsize": (8, 5)})
sns.set_palette('hls', 12)
```

```python
sns.lineplot(data=df, # df name
    x='year', # x -axis column
    y='passengers', # y -axis column
    hue='month' # If not specified, show monthly average and variation
)
plt.show()
```
