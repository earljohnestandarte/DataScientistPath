```python
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.graphics.mosaicplot import mosaic
```

```python
df = sns.load_dataset('titanic')
df.head()
dict1 = {0:'die', 1:'alive'}
dict2 = {'male':'Male', 'female':'Female'}
```

```python
df = df.replace({'survived': dict1}) # Change the value of the column
df = df.replace({'sex': dict2})
df.head()
```

```python
# Graph settings
props = lambda key: {'color': 'lightgreen' if 'alive' in key else 'yellow'}
# Create a graph
mosaic(data = df.sort_values ('sex'),
         index = ['sex', 'survived'],
         properties = props, # Change tile color
         axes_label = True, # Show axis labels
         title='Survival rate by sex' # Graph title
)
plt.show ()
```
