```python
import seaborn as sns
import matplotlib.pyplot as plt
flights = sns.load_dataset('flights')
flights.head()
```

```python
df = flights.pivot_table(index = 'month', # Column to be placed in row position
    columns = 'year', # Columns to be placed in the column positions
    values = 'passengers', # Column to use as data
    aggfunc = 'mean' # data aggregation function 
)
df.head()
```

```python
sns.set_theme(rc ={'figure.figsize':(8,7)})
sns.heatmap(df)
plt.title('Heatmap of Flight', fontsize =20)
plt.show()
```
