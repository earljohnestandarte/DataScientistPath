```python
import plotly.express as px
import webbrowser
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('../spreadsheets/GNI2014.csv')
df.head()
```

```python
fig = px.treemap(
    data_frame = df,
    path=['continent', 'iso3'], # Hierarchical structure of data
    values='population', # Column based on tile area
    color='GNI', # Color temperature reference column
    color_continuous_scale ='Bluyl' # Color Palette
)
```

```python
plt.axis('off') # remove axis ticks
fig.update_layout(margin_t =50, margin_l =25, # margin setting
    margin_r =25, margin_b =25,
    width = 800, # width of the graph (pixels)
    height = 600, # height of the graph (pixels)
    title_text = 'GNI 2014', # Graph title
    title_font_size = 20 # Title font size
)
```

```python
# Saving the graph & displaying it on screen
fig.write_html('treemap.html') # Save result as html file
webbrowser.open('treemap.html') # Check the graph in a web browser 
```
