```python
with open('../txt_files/trump.txt', 'r', encoding='utf-8') as f:
    text = f.read()

print(text[:500])
```

```python
import re, nltk 
from nltk.corpus import stopwords
```

```python

stop_words = set(stopwords.words('english'))
clean_text = ' '.join([w for w in re.sub(r'[^a-z\s]', '', text.lower()).split() if w not in stop_words])
print(clean_text[:500])
```

```python
from collections import Counter
```

```python
freq = Counter(clean_text.split()) 
print(freq.most_common(10))
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
wc = WordCloud(width=800, height=400, background_color ='white').generate(clean_text)
plt.figure(figsize =(15,7))
plt.imshow(wc, interpolation='bilinear') # 'bilinear' : smooth rendering (reduces stair-stepping)
plt.axis('off')
plt.show()
```
