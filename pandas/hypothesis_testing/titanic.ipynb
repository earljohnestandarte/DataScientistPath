```python
import pandas as pd
from scipy import stats
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('titanic')
df
```

```python
# Cross Tabulation
table = pd.crosstab(df['pclass'], df['alive'])
table
```

```python
# calculating survival rate by cabin
cabin1_rate = table['yes'][1] / (table['yes'][1] + table['no'][1])
cabin1_rate
```

```python
cabin2_rate = table['yes'][2] / (table['yes'][2] + table['no'][2])
cabin2_rate
```

```python
cabin3_rate = table['yes'][3] / (table['yes'][3] + table['no'][3])
cabin3_rate
```

```python
chi2_stat, p_val, dof, expected_freq = chi2_contingency(table)
pd.DataFrame(expected_freq, index=table.index, columns=table.columns)
```

```python
print("\n--- Chi-Square Test Results ---")
print(f"Chi-square Statistic: {chi2_stat:.4f}")
print(f"p-value: {p_val}")
print(f"Degrees of Freedom: {dof}")
```
