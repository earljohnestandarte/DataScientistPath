```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv ('../../spreadsheets/wilcoxon_test.csv')
```

```python
# Data Exploration
df.mean () # group-wise average
```

```python
(df['post']-df['pre']).mean() # pre,post Average of differences
```

```python
fig, axes = plt.subplots(nrows =1, ncols=2)
df['pre'].plot.box(grid=False, ax=axes[0])
plt.ylim ([60,100])
df['post'].plot.box(grid=False, ax=axes[1])
plt.show ()
```

```python
stats.wilcoxon (df['pre'], df['post'])
```
