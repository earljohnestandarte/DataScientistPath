```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../spreadsheets/prestige.csv')
df
```

```python
# Check the sample size and average income for each group
df.groupby('type').count()
```

```python
df.groupby('type')['income'].mean()
```

```python
df_sub = df[df['type'].isin(['wc', 'bc'])]
df_sub
```

```python
income_bc = df_sub[df_sub['type'] == 'bc']['income']
income_wc = df_sub[df_sub['type'] == 'wc']['income']
income_bc.mean()
```

```python
income_wc.mean()
```

```python

shapiro_bc = stats.shapiro(income_bc)
shapiro_wc = stats.shapiro(income_wc)
shapiro_bc
```
