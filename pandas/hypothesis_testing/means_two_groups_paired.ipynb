```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv ('../../spreadsheets/paired_ttest.csv')
```

```python
# Data Exploration
df.head ()
```

```python
df[['before', 'after']].mean() # Mean by group
```

```python
(df['after']-df['before']).mean() # Differences of before and after average
```

```python
fig, axes = plt.subplots ( nrows =1, ncols =2)
df['before']. plot.box (grid=False, ax=axes[0])
plt.ylim ([60,100])
df['after']. plot.box (grid=False, ax=axes[1])
plt.show ()
```

```python
# Normality test
stats.shapiro (df['after']-df['before'])
```

```python
# Paired T- test
result = stats.ttest_rel (df['before'], df['after'])
result
```
