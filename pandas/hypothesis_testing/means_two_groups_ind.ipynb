```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv ('../../spreadsheets/ind_ttest.csv')
```

```python
df.head ()
```

```python
df.groupby('group').count() # Sample size by group
```

```python
df.groupby ('group').mean() # group-by-group average
```

```python
df.groupby('group').boxplot(grid=False)
plt.show ()
```

```python
group_1 = df.loc[ df.group =='A','height']
group_2 = df.loc[ df.group =='B','height']
group_1

```

```python
group_2
```

```python
# Normality test
stats.shapiro(group_1)
```

```python
stats.shapiro(group_2)
```

```python
# Test for homoscedasticity
stats.levene(group_1, group_2)
```

```python
# Independent samples T- test
result = stats.ttest_ind(group_1, group_2, equal_var =True)
result
```
