```python
import folium
import os

import displayMap as dm
import pandas as pd
```

```python
df = pd.read_csv('../../spreadsheets/wind.csv')
df.head()
```

```python
df = df.sample(50, random_state = 123)
```

```python
center = [df.lat.mean(), df.lon.mean()]
```

```python
map = folium.Map(location=center, zoom_start =5)
dm.showMap(map)
```

```python
for i in range(len(df)):
    folium.Marker(location=[df.lat.iloc[i], df.lon.iloc[i]],
                       icon= folium.Icon(color='blue', icon='flag')
                       ).add_to(map)
dm.showMap(map)
```

```python
map = folium.Map(location=center, zoom_start =5) # Map without markers
for i in range( len (df)):
    folium.CircleMarker(location=[df.lat.iloc[i], df.lon.iloc[i]],
                        radius=(df.spd.iloc[i]**0.5)*2, # radius of the circle
                        color='red', # color of the circle
                        stroke=False, # no outline
                        fill=True, # inner color of the circle
                        fill_opacity ='50%' # transparency of the inner color of the circle
                       ).add_to(map)
dm.showMap(map)
```
