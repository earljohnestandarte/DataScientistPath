```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
late = pd.Series([5,8,7,9,4,6,12,13,8,6,6,4],
                 index=list(range(1, 13)))
late.plot(
    title = 'Late Student per month',
    xlabel='month',
    ylabel='frequency',
    linestyle='solid'
)
plt.show()
```

```python
late.plot(
    title='Late student per month',
    xlabel='month',
    ylabel='frequency',
    linestyle='dashed',
    marker='o'
)
plt.show()
```

```python
late1 = [5,8,7,9,4,6,12,13,8,6,6,4]
late2 = [4,6,5,8,7,8,10,11,6,5,7,3]
```

```python
dict = {'class_1': late1, 'class_2': late2}
late = pd.DataFrame(dict, index=list(range(1,13)))
late
```

```python
late.plot(title='Late student per month',
          xlabel='month',
          ylabel='frequency',
          marker='o')
plt.legend(
    loc='upper right'
)
plt.show()
```
