```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../spreadsheets/iris.csv')
df
```

```python
dict = {
    'setosa': 'red',
    'versicolor': 'green',
    'virginica': 'blue'
}
colors = list(dict[key] for key in df.species)
colors
```

```python
df.plot.scatter(
    x='petal_length',
    y='petal_width',
    s=30,
    c=colors,
    marker='o'
)
plt.show()
```
