```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../spreadsheets/mtcars.csv')
df
```

```python
df.plot.scatter(x='wt', y='mpg')
plt.show()
```

```python
df.plot.scatter(
    x='wt',
    y='mpg',
    s=4,
    c='red',
    marker='s'
)
plt.show()
```
