```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv('../../spreadsheets/iris.csv')
sw = df.sepal_width
```

```python
z = np.abs(stats.zscore(sw))
outliers = sw[z > 2]
outliers
```

```python
Q1 = sw.quantile(0.25)
Q3 = sw.quantile(0.75)
IQR = Q3 - Q1
```

```python
outliers = sw[(sw < Q1 - IQR*1.5) | (sw > Q3 + IQR*1.5)]
outliers
```

```python
clean = sw.loc[~sw.isin(outliers)]
len(clean)
```
