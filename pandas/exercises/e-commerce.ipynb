```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('../../spreadsheets/Ecommerce_Orders_Delivery.xlsx')
df.head()
```

```python
df.plot()
plt.show()
```

```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]
df['Month'] = pd.Categorical(df['Month'], categories=month_order, ordered=True)
df_sample = df.sample(n=50, random_state=42).dropna(subset=['Month', 'Delivery Days'])

df_sample = df_sample.sort_values('Month')
```

```python
df_sample.plot(
    x="Month",
    y="Delivery Days"
)
plt.show
```

```python
data = df['Payment Method'].value_counts()
labels = df['Payment Method'].unique()
labels

plt.pie(
    data,
    labels=labels,
    autopct='%1.1f%%',   # Show percentages
    startangle=90,       # Rotate start
)
plt.title("Payment Methods")
plt.show()

```
