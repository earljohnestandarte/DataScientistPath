```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
housing_data = pd.read_csv('../../spreadsheets/HousingData.csv')
housing_data
```

```python
house_price = housing_data['MEDV']
house_price
```

```python
quartiles = house_price.quantile([0.25, 0.5, 0.75, 1])
quartiles.plot.bar(
    rot=45
)
plt.show()
```

```python
house_price.plot.box(
    rot=45
)
plt.show()
```

```python
house_price.plot.hist(
    rot=45
)
plt.show()
```
