```python
import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from plotly.offline import iplot, plot
from plotly.subplots import make_subplots

colors = ["#b1e7cd","#854442","#000000","#fff4e6","#3c2f2f",
           "#be9b7b ","#512E5F","#45B39D","#AAB7B8 ","#20B2AA",
           "#FF69B4","#00CED1","#FF7F50","#7FFF00","#DA70D6"]
```

</div>


```python
car = pd.read_csv("/kaggle/input/cars-dataset/cars.csv")
```

```python
#LETS GET SAMPLE FROM THE DATA
car.sample(5)
```

```python
# GET NUM OF COLOMS AND ROWS
print(f"Number of column :{car.shape[1]}\nNumber of rows :{car.shape[0]}")
```

```python
# GET THE SUM OF NULL VALUES
car.isnull().sum()
```

```python
# GET THE INFO 
car.info()
```

```python
pd.DataFrame({'count': car.shape[0],  
              'cardinality': car.nunique()
             })
```

```python
car.describe(exclude=np.number)
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">First Name 🧑</h1>
</div>


```python
top_20_first_name_repeat = car['First Name'].value_counts()[0:20]
```

```python
iplot(px.bar(top_20_first_name_repeat,
             title = 'top 20 first name repeated',
             color_discrete_sequence = colors,
             text_auto = True,
             labels = dict(index = 'First Name',value = 'sum')))
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">Last Name 👴</h1>
</div>


```python
top_20_Last_name_repeat = car['Last Name'].value_counts()[0:20]
```

```python
iplot(px.bar(top_20_Last_name_repeat,
             title = 'top 20 last name repeated',
             color_discrete_sequence = [colors[1]],
             text_auto = True,
             labels = dict(index = 'Last Name',value = 'sum')))
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">Country 🗺️</h1>
</div>


```python
country_counts = car['Country'].value_counts().reset_index()
country_counts.columns = ['Country', 'count']
```

```python
# Plotting the world map based on country occurrences
fig = px.choropleth(locations=country_counts['Country'], locationmode='country names', 
                    color=country_counts['count'], hover_name=country_counts['Country'],
                    title='Occurrences by Country')

fig.update_layout(geo=dict(bgcolor='black'))

fig.show()
```

```python
top_10_country_counts = car['Country'].value_counts()[0:10]
```

```python
iplot(px.bar(top_10_country_counts,
             title = 'top 10 country counts',
             color_discrete_sequence = [colors[3]],
             text_auto = True,
             labels = dict(index = 'Country',value = 'sum')))
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">Car Brand 🚘</h1>
</div>


```python
top_15_car_brands = car['Car Brand'].value_counts().head(15).reset_index()
top_15_car_brands.columns = ['Car Brand', 'Count']

fig = px.scatter(top_15_car_brands,
                 x='Count',
                 y='Car Brand',
                 color='Car Brand',
                 color_discrete_sequence=[colors[4]],
                 size='Count',
                 labels={'Car Brand': 'Car Brand', 'Count': 'Count'},
                 title='Top 15 Car Brand')

fig.update_traces(marker=dict(symbol='circle', line=dict(width=0.5, color='DarkSlateGrey')),
                  selector=dict(mode='markers'))

fig.show()
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">Car Model 🚗</h1>

```python
top_15_car_model = car['Car Model'].value_counts()[0:15]
```

```python
iplot(px.bar(top_15_car_model,
             title = 'top 10 Car Model',
             color_discrete_sequence = [colors[5]],
             text_auto = True,
             labels = dict(index = 'Car Model',value = 'sum')))
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">Car Color </h1>
</div>


```python
sum_car_colors = car['Car Color'].value_counts()
```

```python
fig = px.line(x=sum_car_colors.index, y=sum_car_colors.values, markers=True, title='Counts of Car Colors')
fig.update_xaxes(title='Car Colors')
fig.update_yaxes(title='Count')
fig.show()
```

<div style="background-color: #4b3832; padding: 20px; border-radius: 10px; box-shadow: 5px 5px 10px #888888;">
  <h1 style="color: white; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">Year of Manufacture 📆</h1>
</div>


```python
Year_of_Manufacture = car['Year of Manufacture'].value_counts()
```

```python
fig = px.line(x=Year_of_Manufacture.index, y=Year_of_Manufacture.values, markers=True, title='Counts of Cars Manufactured by Year')
fig.update_xaxes(title='Year of Manufacture', type='category')  
fig.update_yaxes(title='Count')  
fig.update_traces(textposition='top right')  
fig.show()
```

```python
fig = px.histogram(car, x='Year of Manufacture', 
                   title='Year of Manufacture Distribution in Top Hits', 
                   color_discrete_sequence=[colors[7]])  

fig.show()
```

```python
fig = px.pie(names=top_10_credit_card.index,
             values=top_10_credit_card.values,
             title="Top 10 Credit Card Distribution",
             color_discrete_sequence = colors)

fig.update_traces(textinfo='label+percent')
fig.show()
```
