```python
import pandas as pd
from scipy import stats
```

```python
ds = [60, 62, 64, 65, 68, 69]
weight = pd.Series(ds)
ds.append(120)
weight_heavy = pd.Series(ds)
weight
weight_heavy
```

```python
weight.mean()
weight_heavy.mean()
```

```python
weight.median()
weight_heavy.median()
```

```python
stats.trim_mean(weight, 0.2)
stats.trim_mean(weight_heavy, 0.2)
```
