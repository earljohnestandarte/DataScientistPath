```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
colors = pd.Series([2, 3, 2, 1, 1, 2, 2, 1, 3, 2, 1, 3, 2, 1, 2])
```

```python
fd = colors.value_counts()
fd.index = ['red', 'green', 'blue']
fd
```

```python
fd.plot.bar(
    xlabel='Color',
    ylabel='Frequency',
    rot=0,
    title='Favorite Color'
)
plt.show()
```

```python
fd.plot.pie(
    ylabel = '', 
    autopct='%1.0f%%', 
    title='Favorite Color')
plt.show()
```
