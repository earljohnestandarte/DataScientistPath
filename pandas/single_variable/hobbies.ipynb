```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
hobby = pd.Series(['climbing', 'fishing', 'golf', 'swimming', 'climbing', 'climbing',
                     'fishing', 'swimming', 'climbing', 'fishing', 'swimming', 'golf'])

hobby
fig, axes = plt.subplots(nrows = 1, ncols = 2)
hobby_count = hobby.value_counts()
hobby_count.plot.bar(ax=axes[0], rot=45)
hobby_count.plot.pie(ax=axes[1])
plt.suptitle('Multi Graph')
plt.show()
```
